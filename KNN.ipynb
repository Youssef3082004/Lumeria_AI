```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from Column import column
import re
```

```python
places = pd.read_csv("Datasets/Places.csv")
places.head()
```

```python
rows , columns = places.shape
print(f"Places Dataset have {rows} row and {columns} Column")
```

## 🚩 Feature Engineering 
* Make Additional Feature included Places Features 

```python
places["PlaceDescription"] = places["Description"] + " " + places["Location"] + " " + \
" " + places["Historical Era"] + " "+ places["Type Of Site"] + " " + places["Key Features"]
```

## 🚩 Choosing Features that will matter to Text Preprocessing

```python
df = places.loc[ : , ["ID","Name","PlaceDescription"]]
```

## 📍 Lowering Text

```python
df["Cleaned_Description"] = df["PlaceDescription"].apply(lambda x:x.lower())
df
```

```python
desc = column(Column_name="Cleaned_Description",Dataset=df)
```

## 📍 Start Calling **`nltk`** Modules

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

## 📍 Tokenizing Places Description

```python
desc.Update(func=lambda x: word_tokenize(x))
```

## 📍 Removing StopWords

```python
stopwords_ = stopwords.words("english") 
print(f"Number of Stopwords in English = {len(stopwords_)} word")
```

```python
def Remove_Stopwords(Text:str)-> list:
    return [word for word in Text if word not in stopwords_]
```

```python
desc.Update(lambda x:Remove_Stopwords(Text=x))
```

## 📍 Lemmatizing Words

```python
lemma = WordNetLemmatizer()
def lemmatize_word(Text:str) -> list:
    return [lemma.lemmatize(word=word,pos="v") for word in Text]
```

```python
desc.Update(func=lambda x:lemmatize_word(Text=x))
```

## 📍 Trans List to String

```python
desc.Update(func=lambda x:" ".join(x))
```

## 📍 Remove any Punctuation

```python
desc.Update(func=lambda x:re.sub(pattern=r'[\[\]]|[^a-zA-Z\s]',repl=" ",string=x))
```

## 📍 Remove any Single Letter

```python
desc.Update(lambda x:re.sub(pattern=r"\s[a-zA-z]{1}\s",repl=" ",string=x))
```

## 📍 Remove Extra Spaces

```python
desc.Update(lambda x:re.sub(pattern=r"\s+",repl=" ",string=x))
```

## ✔ Start Building Model

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors 
```

## ✔ Vectorize Places Description 

```python
tfidf = TfidfVectorizer(stop_words="english")

tfidf_vctors = tfidf.fit_transform(df["Cleaned_Description"])
```

```python
print(f"Number of Words based in tfidf = {len(tfidf.get_feature_names_out())} Word")
```

## ✔ Get Similaraties between places by KNN

```python
knn = NearestNeighbors(metric="cosine",algorithm="brute",n_neighbors=10)
knn.fit(tfidf_vctors)
```

## 🟢 Build Recommendation System

```python
def get_similar_places(target_id: int, k: int = 6):
    row_idx = target_id - 1     
    distance, indices = knn.kneighbors(tfidf_vctors[row_idx], n_neighbors=k)  

    
    distance = distance.flatten()
    indices = indices.flatten()

    for dist, idx in zip(distance, indices):
        if idx == row_idx:
            continue
            
        row = df.iloc[idx, :].copy()
        row["distance"] = dist
        yield row

pd.DataFrame(get_similar_places(target_id=1,k=10)).sort_values(by="distance")
```

## ✅ Build System with FAISS Lib & Sentence Transformer

```python
import faiss
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer(model_name_or_path="all-MiniLM-L6-v2")
descriptions = model.encode(df["PlaceDescription"]).astype(np.float32)
```

```python
descriptions.shape
```

```python
tfidf_array = tfidf_vctors.toarray().astype(np.float32)
faiss.normalize_L2(descriptions)


dim = descriptions.shape[1]
index = faiss.IndexFlatIP(dim)
index.add(descriptions)
```

```python
np.save("Places_Embeddings.npy", descriptions)
faiss.write_index(index, "faiss_Places_Embeddings.index")
```

```python
tfidf_array = np.load("Places_Embeddings.npy")
index = faiss.read_index("faiss_Places_Embeddings.index")


def get_similar_places(target_id: int, k: int = 6):
    row_idx = target_id - 1     
    query_vector = tfidf_array[row_idx : row_idx+1] 

    distance, indices = index.search(query_vector, k)

    distance = distance.flatten()
    indices = indices.flatten()

    for dist, idx in zip(distance, indices):
        if idx == row_idx:
            continue
            
        row = df.iloc[idx, :].copy()
        row["distance"] = dist
        yield row
        
pd.DataFrame(get_similar_places(target_id=1, k=10)).sort_values(by="distance", ascending=False)
```
